## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Quick evidence from the starter dataset (data/raw/content_refresh_anonymized.csv, 30,000 pages) showing why this lane is worth pursuing: over half the pages (54.2%) are currently trending down, which shows decline is common enough to matter at scale. A stricter combined filter - stale (not updated in 180+ days) AND still getting real visibility (500+ impressions) - narrows this to a focused review list rather than flagging everything as urgent, which is exactly the kind of prioritization a review team needs.

```python
import os, sys, subprocess

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/neha-raniii/flyrank-ml-internship"
REPO_DIR = "flyrank-ml-internship"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)

print("Working dir:", os.getcwd())
assert os.path.exists("data/raw/content_refresh_anonymized.csv"), "starter CSV not found"
print("Starter data found. You're ready.")
```

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

declining_count = (df["trend_direction"] == "down").sum()
declining_pct = (df["trend_direction"] == "down").mean() * 100

stale_and_visible = ((df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)).sum()

print(f"Total pages: {len(df):,}")
print(f"Declining pages: {declining_count:,} ({declining_pct:.1f}% of all pages)")
print(f"Stale but still visible pages (worth reviewing first): {stale_and_visible:,}")
```
